## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
My rule prioritizes content that has high search visibility but low click-through performance. I will give a higher action score to content with more search impressions, lower CTR, and a reasonably good average search position. The goal is to identify content that may benefit from improving its title, description, or search-result presentation.

The action label is IMPROVE_CTR.

The main reason code is HIGH_IMPRESSIONS_LOW_CTR. This reason code means that the content receives substantial search impressions but converts a relatively small share of those impressions into clicks.

The rule is intended as a transparent, directional, decision-support baseline. It does not prove that changing the content will increase clicks.

```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
The ranked queue is created by calculating an action score from search-impression volume, CTR, and average search position. Higher search impressions increase the priority because a possible improvement could affect more search traffic. Lower CTR increases the priority because the content may not be attracting enough clicks. Content with a reasonably good position receives more priority because it is already visible in search results.

The rows are sorted from the highest action score to the lowest action score. Each row includes an action label and a reason code. The ranked queue is written to work/outputs/baseline_action_score.csv.

Only measurements available at the decision moment are used. No future-window outcomes, label-derived fields, or product-generated flags are used.

```
