## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
-- Query 1: Verify the grain
-- Replace entity_id and observation_date with the actual row key.

SELECT
  entity_id,
  observation_date,
  COUNT(*) AS rows_per_key
FROM `PROJECT.DATASET.TABLE`
WHERE month = '2026-03'
GROUP BY entity_id, observation_date
HAVING COUNT(*) > 1
ORDER BY rows_per_key DESC;

-- Query 2: Row count and date span for the slice

SELECT
  COUNT(*) AS row_count,
  MIN(observation_date) AS first_date,
  MAX(observation_date) AS last_date
FROM `PROJECT.DATASET.TABLE`
WHERE month = '2026-03';

-- Query 3: Availability check
-- The task specifically requires IS TRUE.

SELECT
  COUNT(*) AS total_rows,
  COUNTIF(available IS TRUE) AS available_rows,
  COUNTIF(available IS NOT TRUE) AS unavailable_or_unknown_rows
FROM `PROJECT.DATASET.TABLE`
WHERE month = '2026-03';
```
