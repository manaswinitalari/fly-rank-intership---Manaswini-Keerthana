## 2. The question: decision, action, cost of a wrong call

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
Decision: which pages should the content team review this week.
Who acts on it: an SEO/content manager with a fixed weekly review capacity
(e.g. can only realistically check ~50 pages).
Cost of a wrong call: reviewing a page that wasn't actually declining wastes
reviewer time; missing a page that WAS declining means it keeps losing
traffic/rankings unnoticed for another week. Both costs are time and
opportunity cost, not money directly — but a full queue means only the top
of the ranking gets looked at, so a bad ranking has a real cost.
```
